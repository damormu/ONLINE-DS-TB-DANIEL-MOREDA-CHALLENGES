```python
top_5 = df
top_5 = top_5.sort_values("population_total",ascending = False)
top_5
```

```python
fig, ax = plt.subplots(2, 2, figsize = (14, 7))
```

```python
'''
1. Las ciudades más pobladas, con los numeros de las poblaciones (tendrás que usar texto y recordar que para centrarlos se necesita poner el argumento `ha` a "center").
 (Colores: azul, gris, aceituna, marrón, con trasparencia)  
'''
top_5 = df
top_5 = top_5.sort_values("population_total",ascending = False).head(5)

ciudades = top_5["city"].tolist()
valores = top_5["population_total"].abs().tolist()
colores = ["blue","grey","red","olive","brown"]
plt.figure(figsize=(10,10))
plt.bar(ciudades,valores, color = colores ,alpha = 0.3)

barras = plt.bar(ciudades,valores, color = colores ,alpha = 0.3)
for barra in barras:
    yval = barra.get_height()
    plt.text(barra.get_x() + barra.get_width() / 2, yval,
             f'{int(yval):,}',
             ha = "center",
             va = "bottom",
             fontsize =10,
             color = "black")

ax[0,0].set_xlabel("Ciudades")
ax[0,0].set_ylabel("Habitantes (miles)")
ax[0,0].set_title("Ciudades más pobladas de California")
ax[0,0].ticklabel_format( style = "plain", axis = "y")


```

```python
# 2. Histograma del área total ocupada por las ciudades. (50 bins, fuente = "Bradley Hand ITC", tamaño 18, color azul acero o "steelblue")

plt.figure(figsize =(10,6))
ax[0,1].hist(df["area_total_km2"], bins = 50, color = "steelblue", alpha = 0.8, label = "Hist Area");
ax[0,1].set_xlabel("Superfície (KM^2)")
ax[0,1].set_ylabel("Número de ciudades")
ax[0,1].set_title("Ciudades más pobladas de California")
ax[0,1].legend()

#Supongo que bradley ya no es un tipo de fuente aceptado, por eso no sale bien
```

```python
#3. El scatter de área de agua de un ciudad con su poblacion.

plt.figure(figsize=(10,6))
ax[1,0].scatter(df["area_water_sq_mi"], df["population_total"], color = "green", alpha = 0.3);
ax[1,0].set_xlabel("Póblacion por kilometro cuadrado de agua")
ax[1,0].set_ylabel("Población (miles de habitantes)")
ax[1,0].set_title("Área de agua en la ciudad(km^2)")
```

```python
ax[1,1].boxplot(df["elevation_m"], whis = 1.5, vert = False);
ax[1,1].set_xlabel("Altitud (m)")
ax[1,1].set_title("Diagrama de caja de la Altitud de las ciuddades de California")
```
